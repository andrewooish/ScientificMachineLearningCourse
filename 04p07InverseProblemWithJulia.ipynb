# 4.7 Inverse Problem with Julia

```julia
using CairoMakie
using NonlinearSolve
using Zygote
using Optimization
using OptimizationOptimisers
using ForwardDiff
using NonlinearSolve
```

Define the function that we need to solve. 

```julia
f(r,V)=[4π*r[1]-2*V[1]*(1/r[1])*(1/r[1])]
```

---

We will now use the `NonlinearSolve` package and 

```julia
rinit=1.0
prob=NonlinearProblem(f,[rinit],[1.0])
sol = NonlinearSolve.solve(prob)
```

```julia
function loss(V,r)
    prob=NonlinearProblem(f,r,V)
    sol = NonlinearSolve.solve(prob)
    return sum((sol.u[1]-0.6).^2)
end
```

```julia
function my_callback(state, l)
    push!(loss_history, l)
    return false # continues the solver
end

optf = OptimizationFunction(loss, ADTypes.AutoForwardDiff())
prob = OptimizationProblem(optf,[1.0],[1.0])
loss_history = []
sol = solve(prob, Optimisers.Adam(0.01,(0.9,0.99));maxiters=200,callback=my_callback);
```
